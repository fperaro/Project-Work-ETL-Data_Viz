```python
import pandas as pd
from neo4j import GraphDatabase
import json
import pprint
import numpy as np
import geopandas as gpd
from shapely.geometry import Point, Polygon
import h3
import folium
import requests
from geopy.distance import geodesic
from tqdm.notebook import tqdm
from dotenv import load_dotenv
import os

load_dotenv()

NEO4J_URI      = os.getenv("NEO4J_URI")
NEO4J_USER     = os.getenv("NEO4J_USER")
NEO4J_PASSWORD = os.getenv("NEO4J_PASSWORD")
NEO4J_DATABASE = os.getenv("NEO4J_DATABASE")

driver = GraphDatabase.driver(
    NEO4J_URI,
    auth=(NEO4J_USER, NEO4J_PASSWORD)
)
driver.verify_connectivity()
print(f"Connesso a {NEO4J_URI}")

def run(cypher, **params):
    """Cypher → pandas DataFrame."""
    with driver.session(database=NEO4J_DATABASE) as s:
        return pd.DataFrame([r.data() for r in s.run(cypher, **params)])

def run_write(cypher, **params):
    """Run a write query and return its summary counters."""
    with driver.session(database=NEO4J_DATABASE) as s:
        return s.execute_write(
            lambda tx: tx.run(cypher, **params).consume().counters
        )
```

```python
STEP = 0.015

lat_min, lat_max = 42, 48
lon_min, lon_max = 8, 11.5

points = []
for lat in np.arange(lat_min, lat_max, STEP):
    for lon in np.arange(lon_min, lon_max, STEP):
        points.append(Point(lon, lat))

print("punti griglia:", len(points))

regione = gpd.read_file("lombardia.geojson")

valid = []
for p in points:
    if regione.contains(p).any():
        valid.append({"lat": p.y, "lon": p.x})

print("punti validi:", len(valid))
```

```python
comuni_gdf = gpd.read_file("comuni_lombardia.geojson")
comuni_gdf["PRO_COM_T"] = comuni_gdf["PRO_COM_T"].astype(str).str.zfill(6)

pop_df = pd.read_csv("residenti_lombardia_per_fascia.csv", sep=";")
pop_df["Codice comune"] = pop_df["Codice comune"].astype(str).str.zfill(6)


fertile = pop_df[pop_df["Fascia"] == "18-44"][["Codice comune", "Femmine"]].rename(
    columns={"Codice comune": "PRO_COM_T", "Femmine": "donne_fertile"})


comuni_pop = comuni_gdf.merge(fertile, on="PRO_COM_T", how="left")
comuni_pop["donne_fertile"] = comuni_pop["donne_fertile"].fillna(0)

valid_gdf = gpd.GeoDataFrame(
    valid,
    geometry=[Point(p["lon"], p["lat"]) for p in valid],
    crs="EPSG:4326"
)
comuni_pop = comuni_pop.to_crs("EPSG:4326")

joined = gpd.sjoin(
    valid_gdf,
    comuni_pop[["geometry", "PRO_COM_T", "donne_fertile", "COMUNE"]],
    how="left",
    predicate="within"
)

pts_per_comune = joined.groupby("PRO_COM_T").size().rename("n_pts")
joined = joined.join(pts_per_comune, on="PRO_COM_T")
joined["pop_weight"] = (joined["donne_fertile"] / joined["n_pts"]).fillna(0)

pop_weight = joined["pop_weight"].values

print(f"Punti griglia con peso > 0:       {(pop_weight > 0).sum()} / {len(pop_weight)}")
```

```python
time = []
distance = []

Reparto = "OSTETRICIA"

query = """
    MATCH (o:Ospedale)-[:HA_AREA]->(a:AreaSpecialistica)
    WHERE a.Nome = $Reparto
    RETURN o.Coordinate.y AS lat,
           o.Coordinate.x AS lon
"""

hospitals = run(query, Reparto=Reparto)
hosp = np.column_stack([hospitals["lat"], hospitals["lon"]])

with requests.Session() as session:
    for v in tqdm(valid, miniters=10):
        p = (v["lat"], v["lon"])

        nearest = sorted(hosp, key=lambda h: geodesic(p, h).km)[:10]

        coords = [f"{p[1]},{p[0]}"]
        coords.extend(f"{h[1]},{h[0]}" for h in nearest)
        coord_string = ";".join(coords)

        url = f"https://router.project-osrm.org/table/v1/driving/{coord_string}"
        response = session.get(url)
        data = response.json()

        time.append(min(data["durations"][0][1:]) / 60)
        distance.append(geodesic(p, tuple(nearest[0])).km)
```

```python
df = pd.DataFrame({
    "lat": [p["lat"] for p in valid],
    "lon": [p["lon"] for p in valid],
    "dist": [float(d) for d in distance]
})

resolution = 7

df["hex"] = df.apply(
    lambda r: h3.latlng_to_cell(r["lat"], r["lon"], resolution), axis=1
)

hex_df = df.groupby("hex").agg(dist=("dist", "mean")).reset_index()

m = folium.Map(location=[45.5, 9.5], zoom_start=8)

for _, row in hex_df.iterrows():
    boundary = h3.cell_to_boundary(row["hex"])
    poly = [[lat, lon] for lat, lon in boundary]
    folium.Polygon(
        locations=poly, fill=True, fill_opacity=0.7, weight=1, color=None,
        fill_color="red" if row["dist"] > 30 else "orange" if row["dist"] > 15 else "green"
    ).add_to(m)

m.save(f"hex_heatmap_distance_{Reparto}.html")
m
```

```python
df = pd.DataFrame({
    "lat": [p["lat"] for p in valid],
    "lon": [p["lon"] for p in valid],
    "time": [float(t) for t in time]
})

resolution = 7

df["hex"] = df.apply(
    lambda r: h3.latlng_to_cell(r["lat"], r["lon"], resolution), axis=1
)

hex_df = df.groupby("hex").agg(time=("time", "mean")).reset_index()

hex_df["score"] = hex_df["time"].apply(
    lambda d: 2 if d > 30 else 1 if d > 15 else 0
)

m = folium.Map(location=[45.5, 9.5], zoom_start=8)

for _, row in hex_df.iterrows():
    boundary = h3.cell_to_boundary(row["hex"])
    poly = [[lat, lon] for lat, lon in boundary]
    folium.Polygon(
        locations=poly, fill=True, fill_opacity=0.7, weight=1, color=None,
        fill_color="red" if row["time"] > 30 else "orange" if row["time"] > 15 else "green"
    ).add_to(m)

m.save(f"hex_heatmap_time_{Reparto}.html")
m
```

```python
query = '''
MATCH (o:Ospedale)
WHERE NOT EXISTS {
    MATCH (o)-[:HA_AREA]->(:AreaSpecialistica {Nome:$Reparto})
}
RETURN o.Nome AS Nome, o.Coordinate.y AS lat,
       o.Coordinate.x AS lon
'''

noRep = run(query, Reparto=Reparto)
hosp_noRep = np.column_stack([noRep["lat"], noRep["lon"]])
hosp_name  = noRep["Nome"]

print(f"Ospedali candidati (senza {Reparto}): {len(noRep)}")
```

```python
df_aug = df.copy()
df_aug["time_aug"] = None

with requests.Session() as session:
    for idx, row in tqdm(df_aug.iterrows(), total=len(df_aug), miniters=10):
        coords = [f"{row['lon']},{row['lat']}"]
        coords.extend(f"{lon},{lat}" for lat, lon in hosp_noRep)

        url = (
            "https://router.project-osrm.org/table/v1/driving/"
            + ";".join(coords)
        )
        response = session.get(url)
        data = response.json()

        times = np.array(data["durations"][0][1:]) / 60
        df_aug.at[idx, "time_aug"] = dict(zip(hosp_name, times))
```

```python
THRESHOLD_MIN = 30

hosp_name_list = list(hosp_name)

time_aug_matrix = np.array([
    [row["time_aug"].get(name, float("inf")) for name in hosp_name_list]
    for _, row in df_aug.iterrows()
])

time_current = df["time"].values

total_women      = pop_weight.sum()
baseline_unserved = float(np.sum(pop_weight[time_current > THRESHOLD_MIN]))

print(f"Donne 18-44 in Lombardia (dai dati): {total_women:.0f}")
print(f"Donne non servite (>{THRESHOLD_MIN} min) — BASELINE: "
      f"{baseline_unserved:.0f} ({100 * baseline_unserved / total_women:.1f}%)")
print()

score_best = {
    "Name":    None,
    "Unserved": baseline_unserved,
    "Served":   total_women - baseline_unserved,
    "Gain":     0.0,
}

for j, name in enumerate(tqdm(hosp_name_list)):
    best_time = np.minimum(time_current, time_aug_matrix[:, j])
    unserved  = float(np.sum(pop_weight[best_time > THRESHOLD_MIN]))

    if unserved < score_best["Unserved"]:
        score_best["Name"]     = name
        score_best["Unserved"] = unserved
        score_best["Served"]   = total_women - unserved
        score_best["Gain"]     = baseline_unserved - unserved

print()
print(f"Ospedale ottimale:           {score_best['Name']}")
print(f"Donne non servite DOPO:      {score_best['Unserved']:.0f} "
      f"({100 * score_best['Unserved'] / total_women:.1f}%)")
print(f"Donne servite in più:        {score_best['Gain']:.0f}")
```

```python
best_time = np.minimum(
    time_current,
    time_aug_matrix[:, hosp_name_list.index(score_best["Name"])]
)

df_best = pd.DataFrame({
    "lat":  [p["lat"] for p in valid],
    "lon":  [p["lon"] for p in valid],
    "time": best_time
})

df_best["hex"] = df_best.apply(
    lambda r: h3.latlng_to_cell(r["lat"], r["lon"], resolution), axis=1
)

hex_best = df_best.groupby("hex").agg(time=("time", "mean")).reset_index()

m = folium.Map(location=[45.5, 9.5], zoom_start=8)

for _, row in hex_best.iterrows():
    boundary = h3.cell_to_boundary(row["hex"])
    poly = [[lat, lon] for lat, lon in boundary]
    folium.Polygon(
        locations=poly, fill=True, fill_opacity=0.7, weight=1, color=None,
        fill_color="red" if row["time"] > 30 else "orange" if row["time"] > 15 else "green"
    ).add_to(m)

best_row = noRep[noRep["Nome"] == score_best["Name"]].iloc[0]
folium.Marker(
    location=[best_row["lat"], best_row["lon"]],
    icon=folium.Icon(color="blue", icon="plus-sign")
).add_to(m)

m.save(f"hex_heatmap_time_{Reparto}_BEST_popweight.html")
m
```
